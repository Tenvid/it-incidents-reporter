# IT Incidents — Exploratory Analysis

This notebook performs an exploratory analysis, with pandas, of the incidents recorded by the Django application. It reads directly from the project's SQLite database (`src/db.sqlite3`) through a read-only connection, without importing Django or its ORM, so this module stays fully decoupled from the web application. It complements — rather than replaces — the in-app dashboard, offering a quick, on-demand view of the current incident data: totals, open vs. closed counts, and breakdowns by priority and affected equipment.

```python
import sqlite3
from pathlib import Path

import pandas as pd

DB_PATH = Path("src/db.sqlite3").resolve()

connection = sqlite3.connect(f"file:{DB_PATH}?mode=ro", uri=True)

incidents = pd.read_sql_query(
    "SELECT title, equipment, priority, status, is_archived, date FROM incidents_incident",
    connection,
)
connection.close()

incidents.head()
```

## Total incidents

Total number of incidents ever recorded in the system, regardless of status or archive state.

```python
len(incidents)
```

## Open incidents

Incidents that are not closed and have not been archived, whatever their current status (open or in progress).

```python
open_incidents = incidents[
    (incidents["status"] != "closed") & (incidents["is_archived"] == 0)
]
len(open_incidents)
```

## Closed incidents

Incidents whose status has been set to closed.

```python
closed_incidents = incidents[incidents["status"] == "closed"]
len(closed_incidents)
```

## Incidents by priority

Number of incidents for each priority level (low, medium, high).

```python
incidents["priority"].value_counts()
```

## Incidents by equipment

Number of incidents reported for each affected equipment or service.

```python
incidents["equipment"].value_counts()
```
